## Data Pre-processing

#### Required Modules

```python
import os
import pandas as pd
import numpy as np
```

#### Constants

```python
DATA = os.path.join('..', 'data')
```

#### Helper Functions

Method to load data from the `data/raw/` directory.

```python
def load_dataset(filename: str) -> pd.DataFrame:
    """Returns raw dataset."""
    return pd.read_csv(os.path.join(DATA, 'raw', filename))
```

### **1.** Insurance Dataset

```python
INSURANCE = 'insurance.csv'


insurance = load_dataset(INSURANCE)
insurance.head()
```

Transform float `bmi` column into a categorical feature.

```python
if "bmi" in insurance.columns:
    
    bins = [-np.inf, 18.5, 24.9, 29.9, np.inf]
    labels = [
        "0_UNDERWEIGHT", "1_HEALTHY",
        "2_OVERWEIGHT", "3_OBESITY"]

    insurance["BMI_CATEGORY"] = pd.cut(
        insurance["bmi"],
        bins=bins,
        labels=labels)

    insurance = insurance.drop("bmi", axis=1)

insurance.head()
```

Simplify integer `children` column into boolean flag.

```python
if "children" in insurance.columns:
    
    has_children = insurance["children"] >= 1
    insurance["HAS_CHILDREN"] = has_children
    
    insurance = insurance.drop("children", axis=1)

insurance.head()
```

Convert float `charges` column to categorical values (quartiles).

```python
if "charges" in insurance.columns:

    insurance['LABEL'] = pd.qcut(
        insurance['charges'],
        q=5,
        labels=[
            'Q1', 'Q2', 'Q3', 'Q4', 'Q5'])

    insurance = insurance.drop("charges", axis=1)

insurance.head()
```

Standardize column names.

```python
rename_map = {
    "age": "AGE", "sex": "SEX",
    "smoker": "SMOKER", "region": "REGION"}
insurance = insurance.rename(columns=rename_map)

insurance.head()
```

Save data.

```python
filepath = os.path.join(DATA, INSURANCE)
insurance.to_csv(filepath, index=False)
```

### **2.** Personality Dataset

```python
PERSONALITY = 'personality.csv'


personality = load_dataset(PERSONALITY)
personality.head()
```

Transform integer `Time_spent_Alone` column into categorical values.

```python
if "Time_spent_Alone" in personality.columns:
    
    bins = [-np.inf, 3.5, 7.5, np.inf]
    labels = [
        "0_LOW", "1_MEDIUM",
        "2_HIGH"]

    personality["TIME_ALONE"] = pd.cut(
        personality["Time_spent_Alone"],
        bins=bins,
        labels=labels)

    personality = personality.drop("Time_spent_Alone", axis=1)

personality.head()
```

Transform integer `Social_event_attendance` column into categorical values.

```python
if "Social_event_attendance" in personality.columns:
    
    bins = [-np.inf, 3.5, 7.5, np.inf]
    labels = [
        "0_RARE", "1_MODERATE",
        "2_FREQUENT"]

    personality["EVENT_ATTENDANCE"] = pd.cut(
        personality["Social_event_attendance"],
        bins=bins,
        labels=labels)

    personality = personality.drop("Social_event_attendance", axis=1)

personality.head()
```

Standardize column names.

```python
rename_map = {
    "Stage_fear": "STAGE_FEAR",
    "Drained_after_socializing": "SOCIAL_DRAIN",
    "Going_outside": "GOING_OUTSIDE",
    "Friends_circle_size": "FRIEND_CIRCLE",
    "Post_frequency": "POST_FREQUENCY",
    "Personality": "PERSONALITY"}
personality = personality.rename(columns=rename_map)

personality.head()
```

Save data.

```python
filepath = os.path.join(DATA, PERSONALITY)
personality.to_csv(filepath, index=False)
```
